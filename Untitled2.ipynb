```python
# 🎯 CORRECTED COFFEE SHOP ANALYSIS
import numpy as np

daily_sales = np.array([85, 92, 78, 105, 120, 150, 135])  # Cups sold
daily_costs = np.array([3.5, 3.8, 3.2, 4.1, 4.5, 5.2, 4.8])  # Cost PER CUP

print("📊 COFFEE SHOP WEEKLY REPORT:")
print("=" * 40)

# 1. Calculate total weekly revenue (CORRECT!)
total_weekly_revenue = np.multiply(daily_sales, 5)  # $5 per cup
print(f"Daily Revenue: {total_weekly_revenue}")
print(f"Total Weekly Revenue: ${np.sum(total_weekly_revenue)}")

# 2. Calculate total costs first
total_daily_costs = np.multiply(daily_sales, daily_costs)  # sales * cost_per_cup
print(f"Daily Costs: {total_daily_costs}")

# 3. Now calculate profit (revenue - total costs)
total_weekly_profit = np.subtract(total_weekly_revenue, total_daily_costs)
print(f"Daily Profit: {total_weekly_profit}")
print(f"Total Weekly Profit: ${np.sum(total_weekly_profit):.2f}")

# 4. Profit margins (profit/revenue * 100)
profit_margins = np.divide(total_weekly_profit, total_weekly_revenue) * 100
print(f"Daily Profit Margins: {profit_margins}")

# 5. Find best day
best_day_index = np.argmax(profit_margins)  # Index of highest margin
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
print(f"🎯 Best profit margin: {days[best_day_index]} ({profit_margins[best_day_index]:.1f}%)")
```

```python
# 🎯 TASK 2: E-COMMERCE INVENTORY ALERT
product_stock = np.array([15, 8, 45, 2, 23, 12, 30, 5])
min_stock_level = 10

print("\n🚨 INVENTORY ALERT REPORT:")
print("=" * 40)

# 1. Find products needing restock (stock < 10)
need_restock = product_stock < min_stock_level
print(f"Low stock flags: {need_restock}")

# 2. Count how many need restocking
restock_count = np.sum(need_restock)
print(f"Products needing restock: {restock_count}")

# 3. Which specific products? (Get their indices)
low_stock_indices = np.where(need_restock)[0]
print(f"Low stock product indices: {low_stock_indices}")

# 4. Total value at risk (low stock items * $20 each)
low_stock_values = product_stock[need_restock] * 20
total_risk_value = np.sum(low_stock_values)
print(f"Total value at risk: ${total_risk_value}")

# 5. Product with highest stock (for promotions)
highest_stock_index = np.argmax(product_stock)
print(f"Promotion candidate: Product {highest_stock_index} ({product_stock[highest_stock_index]} units)")
```
