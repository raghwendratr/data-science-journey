```python
# DAY 1 - COFFEE BUSINESS ANALYSIS
import pandas as pd
import matplotlib.pyplot as plt

# Create our coffee shop data
data = {
    'date': ['2024-01-01', '2024-01-01', '2024-01-01', '2024-01-02', '2024-01-02', '2024-01-02'],
    'product': ['Espresso', 'Latte', 'Cappuccino', 'Espresso', 'Latte', 'Cappuccino'],
    'region': ['North', 'North', 'North', 'South', 'South', 'South'],
    'sales': [150, 200, 180, 170, 240, 190],
    'cost': [50, 80, 70, 55, 85, 75]
}

df = pd.DataFrame(data)
print("✅ Data loaded! Let's analyze our coffee business:")
print(df)
```

```python
# BUSINESS QUESTION 1: What's our best seller?
print("📊 BUSINESS INSIGHTS:")
best_seller = df.groupby('product')['sales'].sum().idxmax()
print(f"1. Best-selling product: {best_seller}")

# BUSINESS QUESTION 2: Most profitable region?
df['profit'] = df['sales'] - df['cost']  # Calculate profit
best_region = df.groupby('region')['profit'].sum().idxmax()
print(f"2. Most profitable region: {best_region}")

# BUSINESS QUESTION 3: Total revenue?
total_revenue = df['sales'].sum()
print(f"3. Total revenue: ${total_revenue}")
```

```python
# Create a simple profit chart
df.groupby('product')['profit'].sum().plot(kind='bar', color='green')
plt.title('Total Profit by Product')
plt.ylabel('Profit ($)')
plt.show()
```
